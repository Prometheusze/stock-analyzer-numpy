```python
import numpy as np
```

```python
change = np.random.randint(-9,10, (29))
starting_price = 100
prices = np.cumsum(np.insert(change, 0, starting_price))
```

```python
print("Daily Stock Prices:")
for i in range(prices.size):
    print(f"Day {i + 1}: ₹{prices[i]}")
```

```python
def analysis(stock):
    return f"""
average price: {stock.mean()}
maximum price: {stock.max()}
minimum price: {stock.min()}
median price: {np.median(stock)}
standard deviation: {np.std(stock)}
Day of max price: Day {np.argmax(stock) + 1}
Day of min price: Day {np.argmin(stock) + 1}"""
```

```python
print("Daily Price Change:")
for changes in range(change.size):
    print(f"Day {changes + 2}-{changes + 1}: {change[changes]}")
```

```python
def daily_returns(stock):
    price_change = np.diff(stock)
    returns_list = []
    for day in range(stock.size - 1):
        daily_percent = (price_change[day]/stock[day]) * 100
        returns_list.append(daily_percent)
        
    return np.array(returns_list)
        
returns_arr = daily_returns(prices)
```

```python
def day_analysis(stock):
    price_change = np.diff(stock)
    profit = price_change[price_change > 0].size
    loss = price_change[price_change < 0].size
    breakeven = price_change[price_change == 0].size
    win_rate = (profit/price_change.size) * 100
    return f"""
Profitable days: {profit}
Losing days: {loss}
Unchanged days: {breakeven}
Win rate: {win_rate}%"""
```

```python
def trade_days(returns, stock):
    best_day = returns.argmax() + 2
    worst_day = returns.argmin() + 2
    return f"""
Best trading day: Day {best_day}
Worst trading day: Day {worst_day}
Percentage returns on best day: {returns[best_day - 2]}%
Percentage returns on worst day: {returns[worst_day - 2]}%
Price on best day: {stock[best_day - 1]}
Price on worst day: {stock[worst_day - 1]}"""
```

```python
def volatility(returns):
    return_std = np.std(returns)
    if return_std < 1:
        return 'Low Volatility'
    elif 1 <= return_std < 2:
        return 'Medium Volatility'
    else:
        return 'High Volatility'
```

```python
def unusual_days(returns):
    return_dev = np.abs(returns - returns.mean())
    unusual = return_dev > (2 * np.std(returns))
    if np.any(unusual):
        result = []
        return_index = np.where(unusual)[0]
        for y in range(return_index.size):
            result.append(f"Day {return_index[y] + 2}: {returns[return_index[y]]}%")
    else:
        return "No unusually large movements observed"

    return result
```
